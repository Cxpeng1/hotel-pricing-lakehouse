```python
# Hotel Pricing Lakehouse Project - Bronze Ingestion

import pandas as pd
from pathlib import Path
from datetime import datetime
import uuid

# Define paths
RAW_PATH = Path("C:\\Users\\cxpen\\Documents\\JOB\\Portfolios\\hotel-pricing-lakehouse\\data\\raw\\hotel_bookings.csv")
BRONZE_DIR = Path("C:\\Users\\cxpen\\Documents\\JOB\\Portfolios\\hotel-pricing-lakehouse\\data\\bronze")
BRONZE_DIR.mkdir(parents=True, exist_ok=True)

# Read raw data
df_raw = pd.read_csv(RAW_PATH)

# Add technical metadata
batch_id = str(uuid.uuid4())

df_bronze = df_raw.copy()
df_bronze["ingestion_timestamp"] = datetime.now()
df_bronze["source_file_name"] = RAW_PATH.name
df_bronze["batch_id"] = batch_id

# Save Bronze data
BRONZE_PATH = BRONZE_DIR / "hotel_bookings_bronze.csv"
df_bronze.to_csv(BRONZE_PATH, index=False)

print("Bronze ingestion completed.")
print(f"Rows ingested: {len(df_bronze)}")
print(f"Columns: {len(df_bronze.columns)}")
print(f"Batch ID: {batch_id}")
print(f"Bronze file saved to: {BRONZE_PATH}")
```
